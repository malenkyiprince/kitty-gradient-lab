```python
print("୨୧ ── задача 1 ── ୨୧")

w = 5
x = 0.97
y = 8.7

print("♡ прогноз:", w*x)
print("♡ помилка:", (w*x-y)**2)

g = 2*(w*x-y)*x
print("✦ крутизна:", g)

w = w-0.1*g
print("♡ нове w:", w)
print("♡ новий прогноз:", w*x)
print("♡ нова помилка:", (w*x-y)**2)

print("˚₊‧꒰ა стало краще! ໒꒱‧₊˚")
```

```python
print("୨୧ ── задача 3 ── ୨୧")

w = 12
lr = 1

print("♡ початкова помилка:", (w-8)**2)

w = w-lr*2*(w-8)
print("✦ крок 1:", w, (w-8)**2)

w = w-lr*2*(w-8)
print("✦ крок 2:", w, (w-8)**2)

print("♡ lr = 1 — коливання")
print("♡ 0 < lr < 1 — збіжність")
```

```python
print("୨୧ ── задача 4 ── ୨୧")

x = 1
y = 6
w1 = 2
w2 = 2

h = x*w1
p = h*w2
e = (p-y)**2

print("♡ h:", h)
print("♡ прогноз:", p)
print("♡ помилка:", e)

dp = 2*(p-y)
g2 = dp*h
dh = dp*w2
g1 = dh*x

print("✦ вплив прогнозу:", dp)
print("✦ крутизна w2:", g2)
print("✦ вплив h:", dh)
print("✦ крутизна w1:", g1)

print("♡ перевірка w1:", (((w1+0.001)*x*w2-y)**2-e)/0.001)
print("♡ перевірка w2:", ((h*(w2+0.001)-y)**2-e)/0.001)

w1 = w1-0.05*g1
w2 = w2-0.05*g2

print("✦ нова помилка:", (x*w1*w2-y)**2)
print("˚₊‧꒰ა помилка зменшилась! ໒꒱‧₊˚")
```
